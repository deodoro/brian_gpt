```python
import numpy as np
import matplotlib.pyplot as plt

# Define the range of quantity
Q = np.linspace(0.1, 50, 400)

# Define the functions
P_demand = (100 - Q) / 2
MC = 4 * Q
P_supply = (Q + 20) / 3
AC = (100 / Q) + 4

# Create the figure and the axes
fig, ax = plt.subplots()

# Plot the curves
ax.plot(Q, P_demand, label='Demand')
ax.plot(Q, MC, label='Marginal Cost')
ax.plot(Q, P_supply, label='Supply')
ax.plot(Q, AC, label='Average Cost')

# Set the labels and the title
ax.set_xlabel('Quantity')
ax.set_ylabel('Price')
ax.set_title('Economic Curves')

# Show the legend
ax.legend()

# Show the plot
plt.show()
```
